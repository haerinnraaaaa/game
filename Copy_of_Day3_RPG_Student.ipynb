# Part 1 — Convert Your Day 2 Code

## Why anything has to change at all

On Day 1 and Day 2 your game talked to you through `print()`. That worked because you and the code
were in the same place — the terminal.

A web page is somewhere else entirely. Your code runs on a server; the player is in a browser.
When the server calls `print()`, the text lands in a log file the player will never see.

So instead of *printing* a message, your methods now **hand the message back** and let the web app
decide where to put it:

```python
# Day 1 / Day 2 — shout into the room
print(f"{self.name} attacks for {damage} damage!")

# Day 3 — hand the message back
return f"{self.name} attacks for **{damage}** damage!"
```

Same sentence. Different destination.

*(The `**stars**` make text bold on the web page — markdown, same as these notebook cells.)*

---

## The four edits

| # | Change | Why |
|---|---|---|
| 1 | `print(...)` → `return f"..."` | The web page needs the text, not the terminal |
| 2 | `damage = self.crit_check(d)` → `damage, crit = self.crit_check(d)` | `crit_check` can't print any more either, so it now returns **two** things: the damage *and* whether it was a crit |
| 3 | Add `CLASS_NAME` and `EMOJI` at the top of your class | The dropdown and the health bars read these |
| 4 | `other.health -= damage` → `other.health = max(0, other.health - damage)` | Health bars break on negative numbers. `max(0, …)` floors it at zero |

Edit 2 in full:

```python
# Day 2 — crit_check printed "Critical Hit!" itself and returned one number
damage = self.crit_check(damage)

# Day 3 — it returns a PAIR, so you unpack it and write the message yourself
damage, crit = self.crit_check(damage)
msg = f"⚔️ {self.name} uses Heavy Slash for **{damage}** damage!"
return msg + " 🔥 *Critical hit!*" if crit else msg
```

If you forget to unpack, `damage` becomes the tuple `(24, False)` and the next line throws
`TypeError: unsupported operand type(s) for -: 'int' and 'tuple'`. Loud and easy to spot — good.


---
## ⚠️ The sneaky one — `return` ends the method

This catches nearly everybody, and it only bites people whose move prints **more than once**.

`print()` can happen as many times as you like. `return` happens **once, and then the method stops
immediately.** Every line after it is dead.

```python
# ❌ BROKEN — only the first line ever runs
def special_attack(self, other):
    return f"{self.name} focuses..."
    return f"...and strikes for {damage}!"   # never happens
```

### The fix: build the message in a variable, return it at the end

```python
# ✅ WORKS
def special_attack(self, other):
    msg = f"{self.name} focuses..."          # start the message
    msg += f"\n\n...and strikes for **{damage}**!"   # add to it
    return msg                                # hand it over, once
```

`+=` on a string means "stick this on the end". `\n\n` makes a new paragraph on the web page.

### And a bonus trick: the guard clause

When a whole method is wrapped in `if self.mana >= 30:`, converting to returns lets you flip it
around and get rid of a whole level of indentation:

```python
# Day 2 shape                        # Day 3 shape
if self.mana >= 30:                  if self.mana < 30:
    ...everything...                     return "❌ Not enough mana!"
else:                                ...everything, one indent shallower...
    print("Not enough mana!")
```

Bail out early on the failure case, then write the success case flat. Same logic, easier to read.

### AFTER — the Day 3 version

Same classes, same numbers, same logic. Run this one.

As you read, find each of the four edits in the code:
`return` instead of `print` · `damage, crit =` unpacking · `CLASS_NAME` and `EMOJI` ·
`max(0, …)` on health.


````python
import random

# =============================================================================
#  AFTER — THE DAY 3 VERSION
#  Compare every method to its commented twin in the cell above.
# =============================================================================

class Character:
    def __init__(self, name):                     # unchanged — never printed
        self.name       = name
        self.health     = 100
        self.max_health = 100
        self.mana       = 50
        self.max_mana   = 50

    def is_alive(self):                           # unchanged — already returned
        return self.health > 0

    # status() is gone — the health bars in the app show this instead.

    def crit_check(self, damage):
        """EDIT 2: returns TWO things now — (damage, was_it_a_crit)."""
        if random.random() < 0.2:
            return damage * 2, True               # no print — hand back the fact
        return damage, False

    def basic_attack(self, other):
        damage, crit = self.crit_check(random.randint(5, 10))   # EDIT 2: unpack
        other.health = max(0, other.health - damage)            # EDIT 4: floor at 0
        msg = f"{self.name} punches {other.name} for **{damage}** damage!"
        return msg + " 🔥 *Critical hit!*" if crit else msg     # EDIT 1: return

    def heal(self):
        if self.mana >= 10:
            amount      = random.randint(15, 25)
            self.mana  -= 10
            self.health = min(self.health + amount, self.max_health)
            return f"💚 {self.name} heals for **{amount}** HP!"
        return f"❌ {self.name} doesn't have enough mana to heal!"

    def meditate(self):
        regen     = random.randint(10, 20)
        self.mana = min(self.mana + regen, self.max_mana)
        return f"✨ {self.name} meditates and restores **{regen}** mana."


# ── Warrior — the reference conversion ───────────────────────────────────────
class Warrior(Character):
    CLASS_NAME = "Warrior"     # EDIT 3: the dropdown reads this
    EMOJI      = "⚔️"          # EDIT 3: shown next to the health bar

    def special_attack(self, other):
        if self.mana >= 10:
            damage, crit = self.crit_check(random.randint(20, 30))
            self.mana   -= 10
            other.health = max(0, other.health - damage)
            msg = f"⚔️ {self.name} uses **Heavy Slash** for **{damage}** damage!"
            return msg + " 🔥 *Critical hit!*" if crit else msg
        return f"❌ {self.name} doesn't have enough mana for Heavy Slash!"


# ── Mage — the EASY conversion: one print per branch → one return per branch ─
class Mage(Character):
    CLASS_NAME = "Mage"
    EMOJI      = "🔮"

    def special_attack(self, other):
        if self.mana >= 15:
            damage, crit = self.crit_check(random.randint(25, 35))
            self.mana   -= 15
            other.health = max(0, other.health - damage)
            msg = f"🔮 {self.name} casts **Fireball** for **{damage}** damage!"
            return msg + " 🔥 *Critical hit!*" if crit else msg
        return f"❌ {self.name} doesn't have enough mana for Fireball!"


# ── Monk — the HARD conversion: SEVEN prints in one method ──────────────────
#     You cannot swap each print for a return — the first one ends the method.
#     Instead: flavour goes in a VARIABLE, add with +=, return ONCE at the end.
#     Also note the guard clause on line 1 — it removes a level of indentation.
class Monk(Character):
    CLASS_NAME = "Monk"
    EMOJI      = "🧘"

    def __init__(self, name):
        super().__init__(name)
        self.hit = False

    def special_attack(self, other):
        if self.mana < 30:                    # guard clause: fail early, leave
            return f"❌ {self.name} doesn't have enough mana for a great hit!"

        self.mana -= 30

        if self.hit:                          # every use after the first
            damage, crit = self.crit_check(random.randint(25, 50))
            other.health = max(0, other.health - damage)
            msg = f"🧘 Monk {self.name} uses a large attack for **{damage}** damage!"
            return msg + " 🔥 *Critical hit!*" if crit else msg

        self.hit = True                       # first use only
        roll = random.randint(0, 100)

        if   roll == 0:   flavour = "Monk misses!"
        elif roll <  10:  flavour = "Monk's focus is off! Low damage!"
        elif roll <  20:  flavour = "Monk is a bit off on the hit. Close to a miss!"
        elif roll <  70:  flavour = "Monk does good damage, focus is good!"
        elif roll <  90:  flavour = "Monk's focus is perfect! Deals good damage."
        elif roll < 100:  flavour = "Monk does a heavy hit! Aim is to be the best!"
        else:             flavour = "WOW! Monk does a PERFECT hit!!! Most damage done!"

        msg = f"🧘 {self.name} the Monk focuses! Ready to hit!\n\n*{flavour}*"

        if roll > 0:
            damage, crit = self.crit_check(roll)
            other.health = max(0, other.health - damage)
            msg += f"\n\nThe Monk strikes {other.name} for **{damage}** damage!"
            if crit:
                msg += " 🔥 *Critical hit!*"

        return msg                            # ONE return, at the very end


# ── Boss — your Day 2 AI, converted (given) ─────────────────────────────────
#     The scoring logic is byte-for-byte identical to Day 2. Only the four
#     lines at the bottom changed: they now RETURN the message they get back.
class Boss(Character):
    CLASS_NAME = "Boss"
    EMOJI      = "👹"

    def __init__(self):
        super().__init__("Doom Bringer")
        self.health = self.max_health = 150
        self.mana   = self.max_mana   = 40

    def special_move(self, target):
        if self.mana < 20:
            return self.basic_attack(target)     # pass the message straight up

        damage, crit = self.crit_check(random.randint(40, 50))
        self.mana    -= 20
        target.health = max(0, target.health - damage)

        # TWO prints on Day 2 → one message built in two steps
        msg  = "💢 Doom Bringer unleashes **Desolation Wave**!"
        msg += f"\n\n💀 {target.name} is hit for **{damage}** massive damage!"
        if crit:
            msg += " 🔥 *Critical hit!*"
        return msg

    def choose_ai_action(self, target):
        options = []
        if self.health < 60 and self.mana >= 10:
            options.append(("heal",     40 + (60 - self.health)))
        if self.mana < 10:
            options.append(("meditate", 30 - self.mana))
        if self.mana >= 20:
            options.append(("special",  10 + (100 - target.health)))
        options.append(("basic",         5 + random.randint(0, 5)))

        best = max(options, key=lambda x: x[1])[0]   # ← unchanged from Day 2

        if best == "heal":     return self.heal()
        if best == "meditate": return self.meditate()
        if best == "special":  return self.special_move(target)
        return self.basic_attack(target)


class Soldier(Character):
     CLASS_NAME = "Soldier"

     def special_attack(self, other):
         if self.mana >= 15:
             damage = random.randint(25, 35)
             damage = self.crit_check(damage)
             self.mana -= 15
             other.health -= damage
             msg = f"{self.name} launches a Thunder Spear for {damage} damage!"
             return msg + "A direct hit to the nape! Critical hit!*" if crit else msg
         return f"{self.name} doesn't have enough gas for Thunder Spear!"



print("✅ Classes defined. Now run the Conversion Test cell below.")
````

````python
# =============================================================================
#  CONVERSION TEST — plain Python, no web server
# =============================================================================
TEST_CLASS = Soldier          # ← change to YOUR class (no quotes, no brackets)

problems = []

def check(label, ok, hint):
    print(f"{'✅' if ok else '❌'}  {label}")
    if not ok:
        problems.append(hint)

print(f"Testing {TEST_CLASS.__name__}\n" + "-" * 52)

# Edit 3 — the UI attributes
check("Has CLASS_NAME", hasattr(TEST_CLASS, "CLASS_NAME"),
      "Add  CLASS_NAME = \"YourName\"  just under the class line.")
check("Has EMOJI", hasattr(TEST_CLASS, "EMOJI"),
      "Add  EMOJI = \"🔮\"  just under the class line.")

hero  = TEST_CLASS("Testy")
dummy = Character("Dummy")
dummy.health = dummy.max_health = 400

# Edit 1 — returns instead of prints
hero.mana = hero.max_mana
result = hero.special_attack(dummy)
check("special_attack RETURNS something", result is not None,
      "Your method still uses print(). Change print(f\"...\") to return f\"...\".")
check("...and it returns TEXT", isinstance(result, str),
      "Return an f-string, not a number or a tuple.")

# Edit 2 / 4 — the attack actually landed, health floored at 0
check("The target actually lost health", dummy.health < 400,
      "Either  other.health = max(0, other.health - damage)  is missing, or "
      "your move costs more mana than you start with (50).")
dummy.health = 5
hero.mana = hero.max_mana
hero.special_attack(dummy)
check("Health never goes negative", dummy.health >= 0,
      "Wrap it: other.health = max(0, other.health - damage)")

# The move still costs SOMETHING — mana for most classes, HP for a few.
# (Top both bars up first: some moves cost a lot.)
hero.mana   = hero.max_mana
hero.health = hero.max_health
mana_before, hp_before = hero.mana, hero.health
hero.special_attack(dummy)
check("Special attack costs something (mana or HP)",
      hero.mana < mana_before or hero.health < hp_before,
      "Your  self.mana -= COST  (or self.health -= COST) line is missing, "
      "or the cost is higher than you can ever afford.")

# The failure branch must return text too
hero.mana = 0
broke = hero.special_attack(dummy)
check("No-mana branch returns text too", isinstance(broke, str),
      "The else / not-enough-mana branch still prints instead of returning.")

print("-" * 52)
if problems:
    print("⚠️  Fix these before deploying:\n")
    for p in problems:
        print("   •", p)
else:
    print("🎉 All checks passed — your class is ready for the web app.")
    print("   Copy it into Cell 2 at TODO 1.")

# =============================================================================
#  BONUS CHECK — watch a whole fight in plain text
#  Every line below came back as a RETURN VALUE. On the web page these same
#  strings go into the battle log instead of the screen.
# =============================================================================
hero = TEST_CLASS("Hero")
boss = Boss()
turn = 0

while hero.is_alive() and boss.is_alive() and turn < 40:
    turn += 1
    # a simple player: heal when hurt, special when affordable, else meditate
    if hero.health < 40 and hero.mana >= 10:
        line = hero.heal()
    else:
        line = hero.special_attack(boss)
        if line.startswith("❌"):          # couldn't afford it — punch instead
            line = hero.basic_attack(boss)
    print(f"T{turn}  {line}")

    if not boss.is_alive():
        print(f"\n🏆 {hero.name} wins in {turn} turns!")
        break

    print(f"T{turn}  {boss.choose_ai_action(hero)}")

    if not hero.is_alive():
        print(f"\n💀 {boss.name} wins in {turn} turns.")
        break
else:
    print("\n⏱️  Stalemate after 40 turns — nobody could finish it.")
````

### 🧠 The one idea that's different from every program you've written

**Streamlit re-runs your whole script, top to bottom, every single time the player clicks anything.**

Not the one function. The whole file. That's how the page redraws itself.

Which means a normal variable is useless — it gets rebuilt from scratch on every click:

```python
turn = 0          # ❌ back to 0 after every button press
turn += 1
```

Anything that must survive a click goes in **`st.session_state`** — a dictionary Streamlit keeps
between re-runs:

```python
st.session_state["turn"] = 0        # ✅ survives
st.session_state["turn"] += 1
```

In Cell 2 there's a shortcut near the top — `state = st.session_state` — so the rest of the code
just writes `state["turn"]`, `state["player"]`, `state["log"]`.

**This is also why `st.rerun()` exists.** After you change something, you ask Streamlit to redraw
the page immediately instead of waiting.

### ✍️ Text and layout

| Function | What it does |
|---|---|
| `st.title("⚔️ Boss Battle")` | Biggest heading |
| `st.subheader("Choose your hero")` | Smaller heading |
| `st.caption("❤️ HP 80 / 100")` | Small grey text |
| `st.markdown("**bold** and *italic*")` | Any markdown — this is how your battle log renders |
| `st.divider()` | A horizontal line |
| `st.columns(2)` | Splits the page into side-by-side columns |

Columns are used with `with`:

```python
left, right = st.columns(2)
with left:
    st.title("You")
with right:
    st.title("Boss")
```

`st.columns([5, 1, 5])` gives **widths** instead of a count — that's the wide/narrow/wide layout
around the ⚡ in the battle screen.

### 🎛️ Things the player interacts with

| Function | Returns |
|---|---|
| `st.text_input("Hero name")` | Whatever they typed, as a string |
| `st.selectbox("Class", ["Warrior ⚔️"])` | The option they picked, as a string |
| `st.button("Attack")` | `True` **only** on the click that pressed it, `False` every other re-run |

That button behaviour is why the code looks like this:

```python
if st.button("⚔️ Basic Attack"):
    action = "basic"
```

Useful button options:

| Option | Effect |
|---|---|
| `type="primary"` | Makes it the bright, obvious button |
| `use_container_width=True` | Stretches it to fill its column |
| `disabled=(count == 0)` | Greys it out — how the potion button locks at zero |

## Cell 2 — Your Game Code

- 👇 Find **TODOs 2, 3, 4** — uncomment 3 lines to add your class to the UI

````python
%%writefile app.py
import streamlit as st
import random

st.set_page_config(page_title="Boss Battle RPG", page_icon="⚔️", layout="centered")

# =============================================================================
# PART 1 — GAME CLASSES
# =============================================================================
#
# ✏️  ONE SMALL CHANGE from your Day 1 / Day 2 code
# ─────────────────────────────────────────────────
# Your terminal game used print() to show messages.
# Streamlit can't read terminal output — it needs the message as a return value.
#
#   OLD (Day 1):  print(f"{self.name} attacks for {damage} damage!")
#   NEW (Day 3):  return f"{self.name} attacks for **{damage}** damage!"
#
# That's the only change needed!  Everything else stays the same.
#
# Also update crit_check calls:
#   OLD:  damage = self.crit_check(damage)          → single return value
#   NEW:  damage, crit = self.crit_check(damage)    → now returns (damage, bool)
#
# =============================================================================

# ── Base Character Class  (given — no changes needed) ─────────────────────────

class Character:
    def __init__(self, name):
        self.name       = name
        self.health     = 100
        self.mana       = 50
        self.max_health = 100
        self.max_mana   = 50

    def is_alive(self):
        return self.health > 0

    def crit_check(self, damage):
        """Returns (damage, is_crit) tuple — no print, just values."""
        if random.random() < 0.2:
            return damage * 2, True
        return damage, False

    def basic_attack(self, other):
        damage, crit = self.crit_check(random.randint(5, 10))
        other.health  = max(0, other.health - damage)
        msg = f"{self.name} punches {other.name} for **{damage}** damage!"
        return msg + " 🔥 *Critical hit!*" if crit else msg

    def heal(self):
        if self.mana >= 10:
            amount      = random.randint(15, 25)
            self.mana  -= 10
            self.health = min(self.health + amount, self.max_health)
            return f"💚 {self.name} heals for **{amount}** HP!"
        return f"❌ {self.name} doesn't have enough mana to heal!"

    def meditate(self):
        regen     = random.randint(10, 20)
        self.mana = min(self.mana + regen, self.max_mana)
        return f"✨ {self.name} meditates and restores **{regen}** mana."

class Soldier(Character):
    CLASS_NAME = "Soldier"
    EMOJI      = "⚔️"

    def __init__(self, name):
        super().__init__(name)
        self.health = self.max_health = 110  # Higher HP for a frontline warrior
        self.mana = self.max_mana = 50

    def special_attack(self, other):
        if self.mana >= 15:
            self.mana -= 15
            damage, crit = self.crit_check(random.randint(25, 35))
            other.health = max(0, other.health - damage)
            msg = f"💥 {self.name} launches a **Thunder Spear** for **{damage}** damage!"
            return msg + " 🎯 *A direct hit to the nape! Critical damage!*" if crit else msg
        return f"💨 {self.name} doesn't have enough gas for a Thunder Spear!"

class Supporter(Character):
    CLASS_NAME = "Supporter"
    EMOJI      = "🧠"

    def __init__(self, name):
        super().__init__(name)
        self.health = self.max_health = 95   # Slightly lower HP, higher dynamic gameplay
        self.mana = self.max_mana = 60       # More tactical gas capacity

    def special_attack(self, other):
        if self.mana >= 15:
            self.mana -= 15
            damage, crit = self.crit_check(random.randint(20, 30))
            # Strategic distraction forces the boss to lose gas/focus
            other.mana = max(0, other.mana - 10)
            other.health = max(0, other.health - damage)
            msg = f"🧠 {self.name} executes a **Flank Distraction** for **{damage}** damage!"
            return msg + " 🎯 *The diversion completely blinds the enemy! Critical hit!*" if crit else msg
        return f"💨 {self.name} doesn't have enough gas for a Flank Distraction!"

# ─────────────────────────────────
# YOUR CLASS GOES BELOW THIS LINE:
#
# ⚠️ Didn't run the Conversion Test yet? Go back and do it. A class that
#    still uses print() will look fine here and then show blank lines in
#    the battle log, with no error message anywhere you can see it.


# =============================================================================
# 💡 Want to add something new to your game?
# =============================================================================
# This is your chance to expand! Ideas:
#   • Give each class different HP or mana (change self.health in __init__)
#   • Add a third class (Rogue, Archer, Paladin…)
#   • Give your character a new special move
#   • Add a passive ability (e.g. Warrior takes 10% less damage)
# =============================================================================

# ── Boss  (given — your Day 2 AI work, already adapted) ───────────────────────

class Titan(Character):
    CLASS_NAME = "Titan"
    EMOJI      = "👹"

    def __init__(self):
        super().__init__("Armored Titan")
        self.health = self.max_health = 150
        self.mana   = self.max_mana   = 40

    def special_move(self, target):
        if self.mana >= 20:
            damage, crit = self.crit_check(random.randint(40, 50))
            self.mana   -= 20
            target.health = max(0, target.health - damage)
            msg = f"💢 **Armored Titan** unleashes a **Hardened Charge** for **{damage}** damage!"
            return msg + " 🔥 *Critical hit!*" if crit else msg
        return self.basic_attack(target)

    def choose_ai_action(self, target):
        # Attempt special move if low HP and mana allows
        if self.health < 50 and self.mana >= 20:
            return self.special_move(target)
        # Priority-based action selection (your Day 2 logic!)
        options = []
        if self.mana >= 10 and self.health < 60:
            options.append(("heal",     40 + (60 - self.health)))
        if self.mana < 10:
            options.append(("meditate", 30 - self.mana))
        if self.mana >= 20:
            options.append(("special",  10 + (100 - target.health)))
        options.append(("basic",        5  + random.randint(0, 5)))
        best = max(options, key=lambda x: x[1])[0]
        if best == "heal":     return self.heal()
        if best == "meditate": return self.meditate()
        if best == "special":  return self.special_move(target)
        return self.basic_attack(target)

# =============================================================================
# PART 2 — STREAMLIT APP  (all given — this is the new Day 3 content!)
# =============================================================================

def init_state():
    defaults = {
        "screen":  "setup",
        "player":  None,
        "boss":    None,
        "log":     [],
        "turn":    0,
        "bandage": 3,   # bonus feature — each restores 30 HP
    }
    for key, val in defaults.items():
        if key not in st.session_state:
            st.session_state[key] = val

init_state()
state = st.session_state

def log(msg):
    state["log"].append(msg)
    if len(state["log"]) > 30:
        state["log"].pop(0)

def render_setup():
    st.title("⚔️ Boss Battle RPG")
    st.subheader("Choose your hero to face the Armored Titan !")
    st.divider()

    col_left, col_right = st.columns(2)
    with col_left:
        name = st.text_input("Hero name", placeholder="Enter your name…")
    with col_right:
        # 👇 TODO 2 — Add your class name to this list
        #    Example: ["Warrior ⚔️", "Engineer 🔮"]
        cls = st.selectbox("Class", ["Supporter 🧠", "Soldier ⚔️"])

    c1, c2 = st.columns(2)
    with c1:
         st.info("**Supporter 🧠**\n\nHP: 95  ·  Flank Distraction (15 Gas)\n\n_Smart front-line fighter who creates openings for the team to strike._")
    with b2 := c2.container():
        st.info("**Soldier ⚔️**\n\nHP: 110  ·  Blade Slash (15 Gas)\n\n_Balanced frontline warrior with high speed and heavy physical damage._")

    player_name = st.text_input("Enter your Scout's name:", value="Mikasa")
    class_choice = st.selectbox("Select Combat Discipline:", ["Soldier", "Supporter"])

    if st.button("🚨 Advance Out of the Walls!", use_container_width=True):
        if class_choice == "Soldier":
            state["player"] = Soldier(player_name)
        else:
            state["player"] = Supporter(player_name)

        state["boss"]    = Titan()
        state["log"] = ["🚨 **Battle Start!** 🛡️ The Armored Titan rises from the shadows…"]
        state["bandage"] = 3
        state["screen"]  = "battle"
        st.rerun()

def render_battle():
    player = state["player"]
    boss   = state["boss"]

    st.title("🧱 Armored Titan Boss Fight")

    left, mid, right = st.columns([5, 1, 5])
    with left:
        st.markdown(f"## {player.EMOJI}  {player.name}")
        st.caption(player.CLASS_NAME)
        st.caption(f"❤️ HP  {player.health} / {player.max_health}")
        st.progress(player.health / player.max_health)
        st.caption(f"💨 GAS  {player.mana} / {player.max_mana}")
        st.progress(player.mana / player.max_mana)
        st.caption(f"🩹 Bandages: {state['potions']}")
    with mid:
        st.markdown(
            "<div style='text-align:center;padding-top:2rem;font-size:2rem'>⚡</div>",
            unsafe_allow_html=True)
    with right:
        st.markdown(f"## {boss.EMOJI}  {boss.name}")
        st.caption(f"❤️ HP  {boss.health} / {boss.max_health}")
        st.progress(boss.health / boss.max_health)
        st.caption(f"🧱 HARDENING  {boss.mana} / {boss.max_mana}")
        st.progress(boss.mana / boss.max_mana)

    st.divider()
    st.subheader(f"🎮 {player.name}'s Turn — choose an action:")
    b1, b2, b3, b4 = st.columns(4)
    action = None
    with b1:
        if st.button("⚔️ Blade Slash",  use_container_width=True): action = "basic"
    with b2:
        btn_label = "💥 Thunder Spear" if player.CLASS_NAME == "Soldier" else "🧠 Flank Move"
        if st.button(btn_label, use_container_width=True): action = "special"
    with b3:
        if st.button("🔋 Restock Gas",    use_container_width=True): action = "heal"
    with b4:
        if st.button(f"🩹 Bandage ({state['potions']})", use_container_width=True,
                     disabled=(state["potions"] == 0)):               action = "potion"

    if action:
        state["turn"] += 1
        log(f"--- **Round {state['turn']}** ---")

        if action == "basic":
            log(player.basic_attack(boss))
        elif action == "special":
            log(player.special_attack(boss))
        elif action == "heal":
            log(player.meditate())  # Re-linked to fuel recharge
        elif action == "potion":
            state["potions"] -= 1
            player.health = min(player.max_health, player.health + 30)
            log(f"🩹 {player.name} applies a field bandage and patches up **30** HP!")

        if not boss.is_alive():
            log(f"🏆 **Humanity Wins!** {player.name} forced the Armored Titan into retreat!")
            state["screen"] = "victory"
            st.rerun()

        log(boss.choose_ai_action(player))

        if not player.is_alive():
            log(f"💀 **The Armored Titan wins!** {player.name} has fallen in battle…")
            state["screen"] = "defeat"
            st.rerun()

        st.rerun()

    st.divider()
    st.subheader("📜 Tactical Log Feed")
    with st.container(height=220):
        for entry in reversed(state["log"]):
            st.markdown(entry)


def render_victory():
    st.balloons()
    st.title("🏆 Victory for Humanity!")
    st.success(f"**{state['player'].name}** defended the walls from the Armored Titan in {state['turn']} rounds!")
    for entry in state["log"][-8:]:
        st.markdown(entry)
    if st.button("🔄 Next Mission", type="primary"):
        for key in ["screen", "player", "boss", "log", "turn", "potions"]:
            del st.session_state[key]
        st.rerun()


def render_defeat():
    st.title("💀 Mission Failed")
    st.error(f"The Armored Titan broke through! **{state['player'].name}** fell in action… try again!")
    for entry in state["log"][-8:]:
        st.markdown(entry)
    if st.button("🔄 Try Again", type="primary"):
        for key in ["screen", "player", "boss", "log", "turn", "potions"]:
            del st.session_state[key]
        st.rerun()


screens = {
    "setup":   render_setup,
    "battle":  render_battle,
    "victory": render_victory,
    "defeat":  render_defeat,
}
screens[state["screen"]]()


# =============================================================================
# 🎉 BONUS CHALLENGES — try these once your game is working!
# =============================================================================
# 1. Give each class unique starting HP/MP — add __init__ to your class and
#    change self.health and self.mana before calling super().__init__(name)
# 2. Add a "Meditate" button to restore mana during battle
# 3. Add a "Flee" button that jumps straight to the defeat screen
# 4. Add a turn limit — if the player hasn't won in 15 turns, they lose
# 5. Add a third playable class (Rogue? Paladin? Archer?)
# 6. Make the Boss drop loot — show a random item message on victory
# 7. Add background music with st.audio() — find a free royalty-free MP3!
# =============================================================================
````

# 🎒 OPTIONAL — Give the Player an Item Bag

*Game deployed and working? Do this. Otherwise finish Part 2 first.*

Right now the player has one item: a potion, tracked by a single number (`state["potions"]`).
That works for one item. It falls apart the moment you want three.

### The data structure: a dictionary

On Day 2 you used a **list of tuples** for the boss's options. A bag is a different shape of
problem — you need to look up *"how many of THIS item do I have?"* — and that's what a
**dictionary** is for:

```python
bag = {"🧪 Potion": 3, "💎 Ether": 2, "💣 Bomb": 1}

bag["🧪 Potion"]        # → 3        look up by name
bag["🧪 Potion"] -= 1   # → 2        spend one
bag.items()             # → every (name, count) pair, for looping
```

| Structure | Shape | Good for |
|---|---|---|
| list of tuples | `[("heal", 70), ("basic", 8)]` | ranking things — Day 2's AI |
| dictionary | `{"Potion": 3, "Bomb": 1}` | looking things up by name — today's bag |

### Why this is nicer than three separate variables

With `potions`, `ethers` and `bombs` as separate numbers, adding a fourth item means editing five
places. With a dictionary, you add one line to the bag and the buttons build themselves — because
the UI code **loops over the bag** instead of naming each item.
